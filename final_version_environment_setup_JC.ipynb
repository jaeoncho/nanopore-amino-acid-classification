```python
# Record the Python and package versions used for this study.
# Run this notebook in the same environment used to execute
# the final signal-processing and analysis notebooks.

from importlib import metadata
from pathlib import Path
import platform
import sys

import pandas as pd


# ---------------------------------------------------------
# Output directory
# ---------------------------------------------------------

# By default, the files are saved in an "environment"
# folder under the current Jupyter working directory.
OUTPUT_DIR = Path("environment")

# Alternatively, uncomment the following two lines to
# explicitly use the current working directory as the
# project directory.
#
# PROJECT_DIR = Path(".")
# OUTPUT_DIR = PROJECT_DIR / "environment"

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


# ---------------------------------------------------------
# Packages used in the analysis
# ---------------------------------------------------------

REQUIRED_PACKAGES = [
    "numpy",
    "pandas",
    "scipy",
    "matplotlib",
    "seaborn",
    "scikit-learn",
    "imbalanced-learn",
    "joblib",
    "statsmodels",
    "pyabf",
    "plotly",
    "ipykernel",
    "jupyter",
]


# ---------------------------------------------------------
# Collect installed package versions
# ---------------------------------------------------------

package_records = []

for package_name in REQUIRED_PACKAGES:
    try:
        package_version = metadata.version(
            package_name
        )
        package_records.append(
            {
                "package": package_name,
                "version": package_version,
            }
        )
    except metadata.PackageNotFoundError:
        print(
            f"Package not found in this environment: "
            f"{package_name}"
        )

software_versions_df = pd.DataFrame(
    package_records
).sort_values(
    by="package"
).reset_index(
    drop=True
)


# ---------------------------------------------------------
# Save software_versions.csv
# ---------------------------------------------------------

software_versions_path = (
    OUTPUT_DIR / "software_versions.csv"
)

software_versions_df.to_csv(
    software_versions_path,
    index=False,
)


# ---------------------------------------------------------
# Save requirements.txt
# ---------------------------------------------------------

requirements_lines = [
    f"{record['package']}=={record['version']}"
    for record in package_records
]

requirements_path = (
    OUTPUT_DIR / "requirements.txt"
)

requirements_path.write_text(
    "\n".join(requirements_lines) + "\n",
    encoding="utf-8",
)


# ---------------------------------------------------------
# Save environment.yml
# ---------------------------------------------------------

environment_lines = [
    "name: nanopore-analysis",
    "channels:",
    "  - conda-forge",
    "dependencies:",
    f"  - python={platform.python_version()}",
    "  - pip",
    "  - pip:",
]

environment_lines.extend(
    [
        f"      - {requirement}"
        for requirement in requirements_lines
    ]
)

environment_path = (
    OUTPUT_DIR / "environment.yml"
)

environment_path.write_text(
    "\n".join(environment_lines) + "\n",
    encoding="utf-8",
)


# ---------------------------------------------------------
# Save general environment information
# ---------------------------------------------------------

environment_info = {
    "Python version": platform.python_version(),
    "Python executable": sys.executable,
    "Operating system": platform.system(),
    "Operating-system release": platform.release(),
    "Platform": platform.platform(),
}

environment_info_path = (
    OUTPUT_DIR / "environment_info.txt"
)

environment_info_path.write_text(
    "\n".join(
        f"{key}: {value}"
        for key, value in environment_info.items()
    )
    + "\n",
    encoding="utf-8",
)


# ---------------------------------------------------------
# Confirm saved files
# ---------------------------------------------------------

print(
    f"Environment records saved to: "
    f"{OUTPUT_DIR.resolve()}"
)

for output_path in [
    requirements_path,
    environment_path,
    software_versions_path,
    environment_info_path,
]:
    print(f"  - {output_path.name}")

display(software_versions_df)
```
